Suponha que uma doença rara afeta uma fração 𝑝 da população. Um teste diagnós
tico tem sensibilidade 𝑠 (probabilidade de dar positivo dado que a pessoa tem a doença) 
e especificidade 𝑒 (probabilidade de dar negativo dado que a pessoa não tem a doença). 
Chame de 𝐷 o evento “a pessoa tem a doença” e de 𝑇 o evento “o teste deu positivo”.

1. Calcule 𝑃(𝐷|𝑇) para 𝑝 = 1%, 𝑠 = 99%, 𝑒 = 99%.
2. Calcule 𝑃(𝐷|𝑇) para 𝑝 = 0.1%, 𝑠 = 99%, 𝑒 = 95%.
3. Discuta por que, mesmo com um teste considerado “bom”, a probabilidade a 
posteriori pode ser surpreendentemente baixa quando a doença é rara.
4. Qual especificidade mínima para 𝑃(𝐷|𝑇) > 50% dado 𝑠 = 99% e 𝑝 = 0.1%?

$$
D = 1\%
$$

$$
s = (T | D) = 99\%
$$

$$
e = (T^c|D^c) = 99\%
$$

### 1. Calcule 𝑃(𝐷|𝑇) para 𝑝 = 1%, 𝑠 = 99%, 𝑒 = 99%.

Para calcular essa probabilidade, podemos aplicar o teorema de Bayes para obter:

$$
P(D|T) = \frac{P(T|D)P(D)}{P(T)}
$$

Sabemos que a sensibilidade do exame é $s = P(T|D) = 99\%$. Além disso, como a doença acomete 1% da população, então $P(D) = 1\%$

Logo:

$$
P(D|T) = \frac{99\% \cdot 1\%}{P(T)} = \frac{0.99\%}{P(T)}
$$

E agora, como calcular $P(T)$? Podemos utilizar a formulação da **probabilidade total**, conforme

$$
P(A) = \sum_{i=1}^n P(A|B_i)P(B_i)
$$

Aplicando-a:

$$
P(T) = P(T|D) P(D) + P(T|D^c) P(D^c)
$$

Como os pares $(T|D^c), (T^c|D^c)$ e $P(D), P(D^c)$ formam uma partição do espaço amostral cada um, então 

$$
(T|D^c) + (T^c|D^c) = P(D) + P(D^c) = 1
$$

Por consequência,

$$
(T|D^c) = 1 - (T^c|D^c) = 1 - 99\% = 1\% 
$$

Assim como

$$
P(D^c) = 1 - P(D) = 1 - 1\% = 99\% \\
$$

O que resulta em

$$
\begin{aligned}
P(T) &= 99\% \cdot 1\% + 1\% \cdot 99\% \\
     &= 0.99\% + 0.99\% \\
     &= 2 \cdot 0.99\%
\end{aligned}
$$

Retomando o problema inicial, tem-se

Logo:

$$
P(D|T) = \frac{0.99\%}{P(T)} = \frac{0.99\%}{2 \cdot 0.99\%} = 50\%
$$

Portanto, a chance de o paciente estar doente dado que o teste foi positivo é de 50% quando $e = s = 99\%$.

## 2. Calcule 𝑃(𝐷|𝑇) para 𝑝 = 0.1%, 𝑠 = 99%, 𝑒 = 95%.

Nesse caso, aplicamos o mesmo princípio da questão anterior.

$$
\begin{aligned}
P(D|T) &= \frac{P(T|D)P(D)}{\textbf{P(T)}} \\[1.2em]
       &= \frac{99\% \cdot 0.1\%}{\textbf{P(T)}} \\[1.2em]
       &= \frac{0.099\%}{\textbf{P(T)}}
\end{aligned}
$$

Calculando $P(T)$

$$
\textbf{P(T)} = P(T|D) P(D) + \mathbf{P(T|D^c) P(D^c)}
$$

Sendo que

$$
\mathbf{P(T|D^c)} = 1 - P(T^c|D^c) = 1 - 95\% = 5\%
$$

e

$$
\mathbf{P(D^c)} = 1 - P(D) = 1 - 0.1\% = 99.9\%
$$

Portanto,

$$
\mathbf{P(T)} = 99\% \cdot 0.1\% + 5\% \cdot 99.9\% = 0.099\% + 4.995\% = 5.094\%
$$

O que nos traz

$$
P(D|T) = \frac{0.099\%}{P(T)} = \frac{0.99\%}{5.094\%} \approx 1.94\%
$$

Assim, a probabilidade de o paciente cujo teste deu positivo ser de fato doente é de 1,94%, dadas as condições descritas.

## 3. Discuta por que, mesmo com um teste considerado “bom”, a probabilidade a posteriori pode ser surpreendentemente baixa quando a doença é rara.

Em uma perspectiva matemática, isso pode ser abordado da seguinte maneira:

$$
\begin{aligned}
P(D|T) &= \frac{P(T|D)P(D)}{P(T)}  \\[1.2em]
       &= \frac{P(T|D)P(D)}{P(T|D) P(D) + P(T|D^c) P(D^c)} \\[1.2em]
       &= \frac{P(T|D)P(D)}{P(T|D) P(D) + (1 - P(T^c|D^c)) P(D^c)} \\[1.2em]
       &= \frac{s \cdot P(D)}{s \cdot P(D) + (1 - e) \cdot P(D^c)} \\[1.2em]
\end{aligned}
$$



Assim, o numerador é proporcional a $ s \cdot P(D)$, o produto da sensibilidade pela parcela da população acometida pela doença. Para doenças raras, o valor de $P(D)$ é minúsculo. Como $s \in [0, 1]$, mesmo com a melhor sensibilidade possível $s = 1$, o numerador ainda é um número minúsuculo.

Já no denominador, o fator $(1 - e) \cdot P(D^c)$ apresenta dois termos. Conforme a especifidade é aprimorada, $(1 - e) \to 0$, visto que $e \in [0, 1]$. Entretanto, a magnitude da melhora da especifidade pode não ser capaz de superar aquela de $P(D^c)$, dado que $P(D^c) \to 1$ conforme a doença é mais rara. Dessa maneira, apenas o fator $(1 - e) \cdot P(D^c)$ pode ser vezes maior do que o numerador $ s \cdot P(D)$, e quando somando a $s \cdot P(D)$, a maior magnitude do denominador sobre o numerador é reforçada.

## 4. Qual especificidade mínima para 𝑃(𝐷|𝑇) > 50% dado 𝑠 = 99% e 𝑝 = 0.1%?

$$
\begin{aligned}
P(D|T) &= \frac{P(T|D)P(D)}{P(T)}  \\[1.2em]
       &= \frac{P(T|D)P(D)}{P(T|D) P(D) + P(T|D^c) P(D^c)} \\[1.2em]
       &= \frac{P(T|D)P(D)}{P(T|D) P(D) + (1 - P(T^c|D^c)) P(D^c)} \\[1.2em]
\end{aligned}
$$



Recordemos que 
$$
e = P(T^c | D^c)
$$

e que
$$
s = P(T|D)
$$

Logo,


$$
\begin{aligned}
       P(D|T) &= \frac{s \cdot P(D)}{s \cdot P(D) + (1 - e) \cdot P(D^c)} \\[1.2em]
       &= \frac{99\% \cdot 0.1\%}{99\% \cdot 0.1\% + (1 - e) \cdot 99.9\%} \\[1.2em]
       &= \frac{99\% \cdot 0.1\%}{99\% \cdot 0.1\% + (1 - e) \cdot 99.9\%} \\[1.2em]
       &= \frac{0.99\%}{0.99\% + (1 - e) \cdot 99.9\%} \\[1.2em]
       % &\approx \frac{1}{1 + (1 - e) \cdot 100.91} \\[1.2em]
\end{aligned}
$$



Assim, para $P(D|T) > \dfrac{1}{2}$:

$$
\frac{1}{1 + (1 - e) \cdot 100} > \frac{1}{2} \\[1.2em]

1 > \frac{1}{2} \cdot (1 + (1 - e) \cdot 100) \\[1.2em]

2 > 1 + (1 - e) \cdot 100 \\[1.2em]

\frac{1}{100} > 1 - e \\[1.2em]

\frac{1}{100} - 1 > -e \\[1.2em]

e > 99\%

$$

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
min, max = -8, -0.1
lista_raridades = np.logspace(min, max,  num=500)
propriedades_exame = (0.99, 0.9999, 0.999999)

resultados = {chave: [] for chave in propriedades_exame}

for propriedade in propriedades_exame:
    for raridade in lista_raridades:
        numerador = propriedade * raridade
        denominador = numerador + (1 - propriedade) * (1 - raridade)
        resultados[propriedade].append(numerador / denominador)
```

```python
for propriedade in propriedades_exame:
    eixo = sns.scatterplot(
        x=np.log10(lista_raridades), 
        y=np.log10(resultados[propriedade]),
        label=f"s = e = {propriedade}"
    )

plt.hlines(y=np.log10(0.5), xmin= max * 0.95, xmax=min * 1.05, color="red", linestyle="--", label="Limiar de 50%")

eixo.set(
    xlabel="log$_{10}$(Raridade da doença)", 
    ylabel="log$_{10}($Probabildiade a posteriori)",
    title="Variações da probabilidade a posteriori",
    xlim=(min * 1.05, max*0.95)
)

plt.legend()
eixo.invert_xaxis()
```
