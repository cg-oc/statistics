## A Estatística Descritiva

"A estatística descritiva é a ferramenta que transforma dados brutos em informações compreensíveis, permitindo identificar padrões, diferenças e pontos fora do comum de maneira rápida. Ela fornece uma visão clara do cenário analisado, tornando mais fácil interpretar números e tomar decisões fundamentadas." [1]

Dessa forma, esse campo de estudo oferece soluções aos mais diversos cenários, como:

**E-commerce**
- A *média* dos gastos por cliente é capaz de fornecer um insight valioso como uma base para a precificação de produtos/serviços, mas a presença de *outliers* (valores anômalos) pode distanciar demais o cálculo da realidade, o que torna a mediana mais interessante nesses casos.

**Indústria de Software**
- A *moda* emerge de forma muito útil para averiguar dados categóricos, como palavras em avaliações de clientes sobre um aplicativo. Dessa forma, é possível analisar palavras-chave para guiar a equipe de desenvolvimento, como "Muito prático", "Travado", "Lento", "Bom design", e assim por diante [2].

## As 2 classes de dados

Na estatística, destacam-se dois grandes grupos de dados: os qualitativos e os quantitativos, sendo que cada um deles apresenta dois subconjuntos. 

**Qualitativos** *nominais*:
- São dados categóricos que são usados para rotular variáveis sem fornecer um valor quantitativo, sem apresentar um "zero" significativo.
- Exemplos incluem gênero, cor do cabelo e nacionalidade.

**Qualitativos** *ordinais*:
- Representam valores discretos ou unidades ordenadas, sendo que a ordem dos elementos importa. Contudo, eles não são medidas e nem quantidades, o que dispensa a aplicação de operações matemáticas sobre esses dados.
- Por exemplo, ordem de chegada de uma corrida, número de telefone e "pequeno, médio e grande".

**Quantitativos** *contínuos*:
- Expressam quantidades e medidas que podem assumir **qualquer** valor em um intervalo, sem que seja possível *contar* (Não há como deifinir o sucessor de 0 em um intervalo até 1). Assim como com dados quantitativos discretos, operações matemáticas podem ser operadas sobre esse tipo de dado.
- São exemplos válidos: altura, a trajetória de uma partícula, uma força aplicada.

**Quantitativos** *discretos*:
- Expressam quantidades e medidas que tem valores determinados e separados, sem uma medida entre eles.
- Como exemplo, há os estados de energia de um elétron, o número de irmãos e quantas letras há em um alfabeto [3].

## Azeitonas e o Sono

Agora vamos observar a *variável aleatória* "Qual a sua opinião sobre azeitona?"

```python
azeitona = df["Qual a sua opinião sobre azeitona?"]

# Aferindo a mediana
mediana = azeitona.median()
print(f"A mediana do gosto por azeitona é {mediana:.2f}/5")
```

Encontramos que a méida da opinião sobre o gosto de azeitona, de 1 a 5, é 4, o que é acima do ponto central de 3 pontos. Pode-se, portanto, inferir que há uma opinião geral positiva a respeito desse alimento. Na minha cosmovisão, isso é um paladar de terrível mau gosto. 

[**Hipótese**] E se o gosto por azeitona for uma consequência de poucas horas dormidas em média? Isto é, o mau gosto talvez origine de um estado mental mais debilitado em razão da privação do sono (que realmente tem efeitos neurológicos negativos!). Busquemos uma possível correlação.

```python
# Renomeando as colunas para que os nomes fiquem mais curtos (Aprendi com o Gemini)
df.rename(columns={'Qual a sua opinião sobre azeitona?': 'opiniao_azeitona',
                    "Quantas horas (em média) você dorme por noite?": "horas_sono"}, 
                    inplace=True)
```

```python
# Remoção de observações apenas se tiverem NaN em algum dos parâmetros relevanates
#                                            a fim de preservar o máximo de dados.
df_tratado = df.dropna(subset=["opiniao_azeitona", "horas_sono"], axis=0) # (aprendi com Gemini)

# Excluindo valores anômalos de média de horas de sono (que são impossíveis!) 
df_tratado = df_tratado[df_tratado["horas_sono"] <= 24]

# Plotando o gráfico
eixo = sns.histplot(x=df_tratado["horas_sono"], y=df_tratado["opiniao_azeitona"])
eixo.set(ylabel="Opinião sobre azeitona", xlabel="Horas dormidas por noite em média");
```

Bom... essa relação parece balanceada, na verdade. Vamos checar a média de horas de sono para opiniões sobre azeitona menores e maiores que 3 (Considerando 3 como neutro).

```python
# Filtrando DF com diferentes opiniões
nao_gostam_de_azeitona = df_tratado[df_tratado["opiniao_azeitona"] < 3]
gostam_de_azeitona = df_tratado[df_tratado["opiniao_azeitona"] > 3]

# Obtendo média de média de horas de sono
media_sono_gostam_de_azeitona = gostam_de_azeitona["horas_sono"].mean().item()
media_sono_nao_gostam_de_azeitona = nao_gostam_de_azeitona["horas_sono"].mean().item()

print(f"Média agregada de horas de sono de quem gosta de azeitona: {media_sono_gostam_de_azeitona}")
print(f"Média agregada de horas de sono de quem NÃO gosta de azeitona: {media_sono_nao_gostam_de_azeitona}")

print("Razão (Sono médio de apreciadores / Sono médio de quem detesta):",
    f"{round(media_sono_gostam_de_azeitona / media_sono_nao_gostam_de_azeitona, 2)*100}%")
```

Veja! Quanta diferença 3% a menos de sono faz a uma pessoa comum. 

Brincadeiras a parte, essa foi só uma avaliação lúdica dos dados. Claramente, não há qualquer relação entre essas duas variáveis aleatórias, e não há estudos cognitivos sobre essa dependência (pelo menos ainda...).

## Altura e Tamanho do sapato

E se estudarmos uma relação que provavelmente faz mais sentido? Vejamos como a altura e o número do sapato se relacionam.

```python
# Renomeando as colunas
df.rename(columns={'Qual é a sua altura em centímetros?': 'altura (cm)',
                    "Qual é o número de sapato que você calça?": "n_sapato"}, 
                    inplace=True)
```

```python
# Remoção de observações apenas se tiverem NaN em algum dos parâmetros relevantes
df_tratado = df.dropna(subset=["altura (cm)", "n_sapato"], axis=0)

# Filtrando valores anômalos de altura (A pessoa mais alta que já nasceu media 2.74 metros!)
df_tratado = df_tratado[df_tratado["altura (cm)"] <= 274]

# Filtrando valores anômalos de altura (A menor pessoa adulta media 55 cm!)
df_tratado = df_tratado[df_tratado["altura (cm)"] >= 55]

# Para corrigir erros de escala como "1,71" -> "171"
df_tratado["altura (cm)"] = df_tratado["altura (cm)"].replace(",", "")

eixo = sns.scatterplot(y=df_tratado["n_sapato"], x=df_tratado["altura (cm)"])
eixo.set(ylabel="Número do sapato", xlabel="Altura (cm)");
```

Como podemos ver, há realmente uma tendência que relaciona proporcionalmente a altura ao tamanho do sapato! Fica nítido que em média pessoas maiores calçam sapatos maiores, mas é também importante importante salientar como mesmo pessoas de mesma altura tem números de sapato diferentes, e de forma bem variada para cada faixa de altura. Isso decorre de fatores genéticos, nutricionais e de tantos outros que podemos saber ou não, e erros de medida também devem ser levados em consideração!

Enfim, e se avaliarmos somente os dados da altura entre si?

```python
# gera boxplot da altura
eixo = sns.boxplot(df_tratado["altura (cm)"])

print("Estatísticas da altura (cm):")
print(f"Mediana: {df_tratado["altura (cm)"].median()}")
print(f"Média: {round(df_tratado["altura (cm)"].mean().item())}")
```

Nesse caso, o tratamento dos dados fez coincidirem a mediana e a média! 

Se observarmos os dados sobre a altura média de altura de garotos e garotas de 19 anos no Brasil, de acordo com [4], a média calculada pelo censo (tratado) se aproxima muito bem daquela da fonte, de 169 cm. Ou seja, apenas 1 cm de diferença! 

Mas e se não houvesse a etapa de tratamento dos dados? Eis o *boxplot* correspondente:

```python
eixo = sns.boxplot(df["altura (cm)"]) # usando Data Frame sem tratamento
```

Sim, é bizarro como esse gigante - a bolinha no topo do gráfico - atrapalha nossa visualização. Vamos comparar o poder descritivo da média e da mediana na presença de dados anômalos como esse:

```python
print("Estatísticas da altura (cm):")
print(f"Mediana: {df["altura (cm)"].median()}")
print(f"Média: {round(df["altura (cm)"].mean().item())}")
```

Quem diria! Nesse caso onde há uma significativa *skewdness* (assimetria) no gráfico, a mediana realmente é mais precisa para indicar um valor intermediário dos dados. Essa **sensibilidade** é de extrema importância para realizar análises mais aproximadas da realidade e ter um fundamento mais sólido para agir sobre a informação.

E qual será a cor favorita dos alunos da Ilum? Podemos estimá-la a partir da *moda* dos dados, a qual é perfeita para dados qualitativos como a cor de maior preferência de um grupo. 

```python
# Renomeando a coluna referente às cores
coluna_cores: str = (
            "Das cores abaixo, qual é a sua favorita (ou a cor que mais " 
            "gosta se sua cor favorita não estiver na lista)?"
)

df.rename(columns={coluna_cores: "cor_fav"}, inplace=True)
```

## Mês de aniversário 🎉

Em tese, as pessoas que passam para a Ilum não devem ter um mês de aniversário mais favorecido, certo? Aliás, isso é um fator real que afeta atletas! Conforme [5], atletas que nascem nos primeiros meses do ano têm uma chance de 3% a 4% maior de serem selecionados à NBA do que aqueles que nascem em dezembro devido ao regimento de separação de idades. Mas a Ilum é outra categoria de competição, não é? Vejamos os dados!

```python
# Criando cópia do DataFrame para não interferir no original
meses_df = df.copy()
meses_df = meses_df.dropna(
    subset=["Qual o mês do seu aniversário?"], 
    axis=0)

# Série com meses de aniversário e sua frequência de aparição
frequencia_meses = meses_df["Qual o mês do seu aniversário?"].value_counts()

# Plotagem
eixo = sns.barplot(frequencia_meses)
eixo.set_xlabel("Mês", fontweight="bold")
eixo.set_ylabel("Contagem", fontweight="bold")
eixo.set_title("Mês de aniversário de estudantes da Ilum (Censo Ilum - 2026)")

# Label da frequência sobre cada barra correspodnente
eixo.bar_label(eixo.containers[0], fontsize=10);
```

Uau. Ok, eu não esperava por essa! Aparentemente, pessoas nascidas nos meses de maio (a moda), julho e outubro têm alguma espécie de vantagem sobre aqueles nascidos nos meses de setembro, dezembro e março. Ou talvez simplesmente houve mais nascimentos nesses meses? É interessante checar a Tabela 2680 do Sistema IBGE de Recuperação Automática (SIDRA) [6], que trata do mês de nascimento de pessoas entre 2003 e 2024. Como sou um aluno da mais nova turma da Ilum, posso afirmar que a maioria dos mais jovens nasceram entre 2007 a 2008, e por isso limitei os dados do IBGE para o intervalo de 2003 a 2008.

OBS: A tabela importada do site do SIDRA não é bem ajustada para os padrões de data frame do `pandas`, então pedi ao Gemini para que consertasse esse problema.

```python
# ---------- Feito pelo Gemini ----------
# O parâmetro header=[0, 1] diz ao Pandas que as linhas 0 e 1 formam o título das colunas
# (Ajuste os números caso haja linhas de texto inútil do IBGE antes do cabeçalho, ex: header=[3, 4])
sidra_df = pd.read_excel("sidra2680.xlsx", header=[0, 1])

# Agora, em vez de "Unnamed", suas colunas serão tuplas, como ('2003', 'Janeiro')
# Para deixar no formato "Tidy" (uma linha por observação), usamos o .stack():
df_limpo = sidra_df.stack(level=[0, 1], future_stack=True).reset_index()

# Renomeando as colunas para ficar bonito
df_limpo.columns = ["Índice_Original", "Ano", "Mês", "Nascimentos"]
df_limpo = df_limpo.drop(columns=["Índice_Original"])

# Agrupa os dados pela coluna 'Mês', seleciona a coluna 'Nascimentos' e soma tudo
soma_por_mes = df_limpo.groupby("Mês")["Nascimentos"].sum().reset_index()

# Definindo a ordem correta dos meses
ordem_meses = ["Janeiro", "Fevereiro", "Março", "Abril", "Maio", "Junho", 
               "Julho", "Agosto", "Setembro", "Outubro", "Novembro", "Dezembro"]

# Avisando ao Pandas que essa coluna tem uma ordem específica
df_limpo["Mês"] = pd.Categorical(df_limpo["Mês"], categories=ordem_meses, ordered=True);

# Agora, quando você fizer o groupby ou o sns.barplot, vai sair certinho de Jan a Dez!
soma_por_mes = df_limpo.groupby("Mês", observed=False)["Nascimentos"].sum().reset_index();
```

Visualizemos ambos os gráficos lado a lado, supondo que o mês de aniversário equivala ao mês de nascimento.

```python
# ---------- Feito com auxílio do Gemini ----------
# 1. Cria a figura com 1 linha e 2 colunas
fig, eixos = plt.subplots(1, 2, figsize=(12, 5))

# Plotagem dos dados do IBGE
sns.barplot(
    x=soma_por_mes.index,
    y = soma_por_mes["Nascimentos"], 
    ax=eixos[0], 
    palette='Blues_d',
    hue=soma_por_mes.index,
    legend=False
    )
eixos[0].set_xlabel("Mês", fontweight="bold")
eixos[0].set_ylabel("Contagem", fontweight="bold")
eixos[0].set_title("Mês de nascimento entre 2003 - 2008 no Brasil (IBGE)");

# Plotagem dos dados do Censo Ilum
sns.barplot(
    x=frequencia_meses.index,
    y = frequencia_meses.values,
    hue=frequencia_meses.index,
    ax=eixos[1],
    palette='Oranges_d',
    legend=False
)

eixos[1].set_xlabel("Mês", fontweight="bold")
eixos[1].set_ylabel("Contagem", fontweight="bold")
eixos[1].set_title("Mês de nascimento de estudantes da Ilum (Censo Ilum - 2026)")

# 4. Ajusta o espaçamento para evitar sobreposição de títulos/rótulos
plt.tight_layout()
plt.show();
```

Pois bem, a discrepância de mais que o dobro de vezes observada entre os aniversariantes em maio e março para os estudantes da Ilum não pode ser explicada pela demografia brasileira, uma vez que há uma mínima diferença entre os nascimentos nesses meses para essa faixa etária. O mesmo vale para os demais períodos.

De acordo com [7], a Lei dos Pequenos Números é um viés cognitivo em que um pesquisador tem alta confiança sobre a validade de conclusões baseadas em uma amostra com poucas observações, sendo que ele costuma atribuir simples ocorrências de variabilidade de dados a padrões, e espera uma reprodutibilidade alta demais para resutados significantes em outras amostras. Portanto, nos resta apenas admitir como a nossa amostragem sobre os alunos da Ilum é pequena demais (apenas 137 pessoas), o que pode gerar falsos padrões a serem observados.
