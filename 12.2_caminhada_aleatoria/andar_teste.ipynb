```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Frequências e defasagem (Razão racional 3:2 gera ciclo fechado)
fx, fy = 3.0, 2.0  # Frequências nos eixos X e Y
delta = np.pi / 4  # Defasagem de fase (45 graus)

# 2. Intervalo de tempo cobrindo um ciclo completo
t = np.linspace(0, 2 * np.pi, 1000)

# 3. Equações paramétricas de posição
x = np.sin(fx * t + delta)
y = np.sin(fy * t)

# 4. Gráfico da trajetória
plt.figure(figsize=(6, 6))
plt.plot(x, y, color="purple", label="Trajetória Recorrente")
plt.scatter(
    [x[0]],
    [y[0]],
    color="red",
    zorder=5,
    label=f"Ponto Inicial (Retorno em t=2π)",
)
plt.title(f"Oscilador 2D - Curva de Lissajous ({int(fx)}:{int(fy)})")
plt.xlabel("Posição X")
plt.ylabel("Posição Y")
plt.grid(True)
plt.legend()
plt.show()
```
