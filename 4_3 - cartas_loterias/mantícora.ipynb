### 1. Qual a probabilidade de sortear duas cartas vermelhas em um baralho convencional (52 cartas)?

    <figcaption> <em> <b>Naipes das cartas de baralho.</b> Fonte: http://fabioblogging.blogspot.com/2015/07/origem-dos-naipes-do-baralho.html</em></figcaption>

Um baralho convencional possui 52 cartas dividas igualmente em 4 naipes, cujas cores são metade em vermelho e a outra metade em preto. Assim, há 26 cartas vermelhas ao total. Supondo que o sorteio seja conduzido *sem reposição*, isto é, que uma carta sorteada não retorne ao baralho, o cálculo da probabilidade segue da seguinte maneira:

- Primeira carta vermelha:

A probabilidade de sortear uma carta vermelha é dada pela razão entre número de cartas vermelhas no baralho e o número de todas as cartas no baralho. Há 26 cartas vermelhas e 52 cartas na totalidade. Assim, a probabilidade de que a primeira carta sorteade seja vermelha é

$$
\dfrac{26}{52} = 0,5
$$ 

- Segunda carta vermelha:

Agora, com uma carta vermelha a menos e, portanto, uma a menos no baralho, também, podemos calcular a probabilidade desse evento ao avaliar

$$
\dfrac{25}{51} \approx 0.49 = 49\%
$$ 


Dessa maneira, a probabilidade de que as duas primeiras cartas sorteadas sejam vermelhas - $\text{P}(A)$ - se dará pelo produto das duas probabilidades calculadas:

$$
\text{P}(A) = \dfrac{26}{52} \cdot \dfrac{25}{51} \approx 0.2451 = 24,51\%
$$

### 2. Qual a probabilidade de sortear 5 cartas do mesmo naipe em um baralho convencional (52 cartas)?



Como não definimos um naipe específico para ser sorteado, o primeiro sorteio servirá apenas para definir qual naipe deverá ser buscado, ou seja, qualquer carta retirada satisfará o evento desejado. Portanto, a probabilidade desse evento específico é de 100% (sempre será sorteada uma carta com algum naipe). 

Já a probabilidade de retirar uma carta de mesmo naipe da primeira para os sorteios seguintes deve levar em consideração quantas cartas já haviam sido removidas e quantas cartas de mesmo naipe ainda sobram. Para cumprir com nosso evento desejado, cada sorteio deve sacar uma carta de mesmo naipe. Assim, a probabilidade para o nosso evento desejado, $\text{P}(B)$, será 

$$
\text{P}(B) = \dfrac{52}{52} \cdot \dfrac{12}{51} \cdot \dfrac{11}{50} \cdot \dfrac{10}{49} \cdot \dfrac{9}{48}
\approx 0,0019 = 0,19\%

$$




Em notação em fatorial, isso também pode ser escrito como:

$$
\text{P}(B) = \dfrac{12}{51} \cdot \dfrac{11}{50} \cdot \dfrac{10}{49} \cdot \dfrac{9}{48} \cdot \dfrac{8!}{8!} \cdot \dfrac{47!}{47!} = \dfrac{12! \cdot 47!}{8! \cdot 51!}
$$

### 3. Qual a probabilidade de se ganhar na loteria federal comprando apenas um bilhete de 6 números? 
A loteria federal sorteia 6 números inteiros diferentes no intervalo de [1,60]. Um bilhete de 6 números te permite escolher 6 números diferentes dentro do intervalo. Ganhar, neste caso, significa que você acertou os 6 números 
sorteados.

<div style="width: 100%; text-align: center; margin-bottom: 20px;">
    <img src="apostas.jpg" alt="Bilhetes de loteria" style="max-width: 60%;"></img>
    <figcaption> <em> <b>Bilhetes de loteria.</b> Fonte: https://www.dci.com.br/financas/loterias/lotofacil/confira-o-resultado-da-lotofacil-mega-sena-quina-e-loteria-federal-deste-domingo-26-7/</em></figcaption>
</div>

Nesse caso, $\Omega$ representa o conjunto de todas as sequências de seis números inteiros de 1 a 60, sem reposição e sem que a ordem importe. Então, o espaço amostral correpondente pode ser calculado a partir da *combinatória* 

$$
\Omega = C^6_{60} = \dfrac{60!}{(60 - 6)!\cdot(6)!} = \dfrac{60!}{54! \cdot 6!} = \dfrac{60 \cdot 59 \cdot ... \cdot 55 \cdot 54!}{54! \cdot 6!} = \dfrac{60 \cdot 59 \cdot ... \cdot 55}{6!} = 50.063.860
$$

Um único bilhete representa uma única combinação de números, então a probabilidade do evento estipulado é de um em cinquenta milhões, sessenta e três mil e oitocentos e sessenta! Ou seja a probabilidade de ganhar nessa loteria com apenas um bilhete ($\text{P}(C)$) é 

$$
\text{P}(C) = \dfrac{1}{50.063.860} \approx 2 \cdot 10^{-8} = 2 \cdot 10^{-6}\%
$$
